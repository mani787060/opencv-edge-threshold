```python
import cv2
import numpy as np
```

```python
# using 'Canny_edge_detection' method :-

# converting your image first into 'grayscale' 
image = cv2.imread("rose image.jpg", cv2.IMREAD_GRAYSCALE)

# apply Canny edge detection method:-
edges = cv2.Canny(image, 50, 150)

# printing an image :-

cv2.imshow("Original image", image)
cv2.imshow("Edges", edges)

cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
# thresholding of an image :-

image = cv2.imread("nature image.jpg", cv2.IMREAD_GRAYSCALE)


# applying for thresholding of an image

thres_value = 120
max_value = 255

ret, thres_img = cv2.threshold(image, thres_value, max_value, cv2.THRESH_BINARY)

# printing an image
cv2.imshow("Original image", image)
cv2.imshow("Threshold image", thres_img)

cv2.waitKey(0)
cv2.destroyAllWindows()
```

if we set here the value of (threshold_value=120),
then if any pixel present in an image having value less than 120 becomes completely black(i.e., 0)
And, if any pixel present in an image having value more than 120 becomes completely white(i.e., 255)

<h2> Bitwise operations </h2>

```python
"""
1 -> cv2.bitwise_and(img1, img2)
2 -> cv2.bitwise_or(img1, img2)
3 -> cv2.bitwise_not(img1)

Note that :-
* img1 and img2 are always having same 'height' and 'width'
** use only black and white images (not working proper on colored images) 

"""
```

```python
img1 = np.zeros((300,300), dtype="uint8")
img2 = np.zeros((300,300), dtype="uint8")

# creating a circle :-
cv2.circle(img1, (150,150), 100, 255, -1)

# creating a rectangle :-
cv2.rectangle(img2, (100,100), (250,250), 255, -1)

# performing all operations one by one :-
bitwise_and = cv2.bitwise_and(img1, img2)
bitwise_or = cv2.bitwise_or(img1, img2)
bitwise_not = cv2.bitwise_not(img1)


cv2.imshow("Circle", img1)
cv2.imshow("Rectangle", img2)
cv2.imshow("AND", bitwise_and)
cv2.imshow("OR", bitwise_or)
cv2.imshow("NOT", bitwise_not)

cv2.waitKey(0)
cv2.destroyAllWindows()
```
